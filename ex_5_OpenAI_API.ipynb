**Traccia:** Scrivi un prompt che generi domande di comprensione su un breve testo.

**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
# Inizializza il client di OpenAI
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()  # legge OPENAI_API_KEY dal file .env
client = OpenAI()
```

```python
input_text = input("Enter text: ")

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0.5,
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un insegnante. Dato un testo, crea 5 domande di comprensione "
                "numerate, nella lingua del testo, con difficoltà crescente: "
                "le prime su fatti espliciti, le ultime su inferenze e significato. "
                "Ogni domanda deve avere risposta basata solo sul testo. "
                "Dopo le domande aggiungi una sezione 'Risposte' con una risposta breve per ciascuna."
            ),
        },
        {"role": "user", "content": input_text},
    ],
)

print(completion.choices[0].message.content)
```
