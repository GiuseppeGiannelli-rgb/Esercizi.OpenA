# Esercizio 3

**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
# Inizializza il client di OpenAI
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()  # legge OPENAI_API_KEY dal file .env
client = OpenAI()
```

```python
input_text = input("Enter text: ")

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0.3,
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un giornalista esperto di sintesi. Riassumi l'articolo in un unico paragrafo "
                "di massimo 255 caratteri, spazi inclusi. Includi solo i fatti essenziali "
                "(chi, cosa, quando, dove, perché), senza opinioni. "
                "Scrivi nella stessa lingua dell'articolo e restituisci solo il riassunto."
            ),
        },
        {"role": "user", "content": input_text},
    ],
)

summary = completion.choices[0].message.content.strip()

# Il modello non conta i caratteri in modo esatto: verifica e taglia se necessario
if len(summary) > 255:
    summary = summary[:252].rsplit(" ", 1)[0] + "..."

print(summary)
print(f"\n({len(summary)} caratteri)")
```
