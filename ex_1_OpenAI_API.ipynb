# Esercizio 1

**Traccia:** Scrivi un prompt che traduca un testo dall'italiano all'inglese e viceversa. Usa il prompt per tradurre una breve descrizione di un prodotto.

**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python
# Inizializza il client di OpenAI
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()  # legge OPENAI_API_KEY dal file .env
client = OpenAI()
```

```python
input_text = input("Enter text: ")
# Esempio: "Cuffie wireless con cancellazione attiva del rumore e 30 ore di autonomia."

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0.3,
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un traduttore professionista italiano-inglese specializzato in descrizioni di prodotto. "
                "Rileva la lingua del testo: se è in italiano traducilo in inglese, se è in inglese traducilo in italiano. "
                "Mantieni tono, formattazione e terminologia tecnica; rendi il testo naturale per un madrelingua. "
                "Restituisci solo la traduzione, senza commenti."
            ),
        },
        {"role": "user", "content": input_text},
    ],
)

print(completion.choices[0].message.content)
```
